```python
grades = []
for i in range(5):
    grade = int(input(f"Enter grade for student {i+1}: "))
    grades.append(grade)
```

```python
total_sum = sum(grades)
print("\nTotal sum of grades:", total_sum)
```

```python
selected = grades[0:5:2]
selected.sort()
print("Selected grades (sorted):", selected)
print("Length of selected list:", len(selected))
print("Number of occurrences of score 5:", selected.count(5))
```

```python
fruits = ("apple", "banana", "orange", "mango", "grape")
print("Original tuple:", fruits)
```

```python
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])
```

```python
new_fruits = ("apple", "strawberry") + fruits[2:]
print("Updated tuple:", new_fruits)
```

```python
extra = ("kiwi", "pineapple")
combined = new_fruits + extra
print("After adding more fruits:", combined)
```

```python
tuple1 = combined[:3]
tuple2 = combined[3:6]
print("First split tuple:", tuple1)
print("Second split tuple:", tuple2)
```

```python
final_tuple = tuple1 + tuple2 + fruits
print("Final combined tuple:", final_tuple)
print("Length of final tuple:", len(final_tuple))
```

```python
poem = """Some say the world will end in fire,
Some say in ice.
From what I’ve tasted of desire
I hold with those who favor fire.
But if it had to perish twice,
I think I know enough of hate
To say that for destruction ice
Is also great
And would suffice."""

new_poem = """Some say life is but a dream,
Some say it's a test.
From what I've seen and what I deem,
I side with those who see it as a quest.

But if it had to end today,
I think I know enough of love,
To say that though it fades away,
It's still what we are made of."""
```

```python
import string
def clean_text(text):
    text = text.lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    return text.split()
```

```python
set1 = set(clean_text(poem))
set2 = set(clean_text(new_poem))
print("Unique words in poem 1:", len(set1))
print("Unique words in poem 2:", len(set2))
```

```python
unique_to_poem1 = set1 - set2
print("\nWords only in poem 1:", unique_to_poem1)
```

```python
unique_to_poem2 = set2 - set1
print("\nWords only in poem 2:", unique_to_poem2)
```

```python
common_words = sorted(set1 & set2)
print("\nWords in both poems (alphabetical):", common_words)
```
